```python
# Pandas Series: 

# A series is one dimensional labeled array capable of holding any data type. The axis labels are collectively called the index 
# Arrays are row wise but series in pandas are column wise. 
```

```python
# Creating Series:  

# There are multiple ways to create a pandas series
```

```python
import numpy as np 
import pandas as pd 

labels = ['a' , 'b' , 'c'] 
mylist = [10 , 20 , 30]  
arr = np.array([10,20,30])
d = {1:10 , 2:20 , 3:30} 

pd.Series(mylist , index=labels)
```

```python
pd.Series(arr) # Series can only have 1-d array and if you try to add 2-d array inside a series this will throw an error
```

```python
pd.Series(d) # When you create a series using a dictionary of python it will take its index by itself as the key property 
```
