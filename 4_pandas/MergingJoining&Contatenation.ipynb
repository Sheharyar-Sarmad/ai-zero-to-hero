```python
import numpy as np 
import pandas as pd 
```

```python
# data frames for working: 

employees = pd.DataFrame({
    'employee_id': [1 , 2 , 3 , 4 , 5], 
    'name': ['John' , 'Anna' , 'Peter' , 'Linda' , 'Bob'],
    'department': ['HR' , 'IT' , 'Finance' , 'IT' , 'HR']
})

# Data Frame two for salary information: 

salaries = pd.DataFrame({
    'employee_id': [1 , 2 , 3 , 6 , 7],
    'salary': [60000 , 80000 , 65000 , 70000 , 90000],
    'bonus': [5000 , 10000 , 7000 , 8000 , 12000]
})
```

```python
salaries
```

```python
employees

# Now if you want to merge two(2) data frames then do it

mergedId = pd.merge(employees, salaries, on="employee_id", how="outer")

# how="outer"
# It keeps ALL rows from BOTH DataFrames (employees + salaries).
# If employee_id matches in both tables → rows are merged.
# If employee_id exists in only one table → still kept, but missing values become NaN.
# Basically it performs a FULL OUTER JOIN (union of both tables).

mergedId = pd.merge(employees , salaries , on="employee_id" , how="left")
mergedId
```

```python
# Concatenation of two data frames:

# data: 

df1 = pd.DataFrame({
    'A': ['A0' , 'A1' , 'A2'],
    'B': ['B0' , 'B1' , 'B2'],
    'C': ['C0' , 'C1' , 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3' , 'A4' , 'A5'], 
    'B': ['B3' , 'B4' , 'B5'] ,
    'C': ['C3' , 'C4' , 'C5']
})
```

```python
# Concatenation according to columns

pd.concat([df1 , df2])
```

```python
# concatenation according to rows

pd.concat([df1 , df2] , axis=1)
```

```python
# **Joining of two(2) data frames


# data 
df1 = pd.DataFrame({
    'name': ["Alice" , "Bob" , "Charlie"]
} , index=[1,2,3])

df2 = pd.DataFrame({
    'score': [85 , 90 , 75]
} , index=[2,3,4])
```

```python
df1.join(df2 , how="outer")
```
