```python
# Group by and aggregation: 

# Group By:
# Group by is a process of dividing data into groups based on a common value or key, so that each group contains similar items together.

# Aggregation:
# Aggregation is the process of performing a mathematical or statistical operation on each group of data,
# such as sum, mean, count, max, or min, to get a single summarized result for each group.
```

```python
import numpy as np 
import pandas as pd 
```

```python
# data: 

data = {
    'Category': ['A','B','A','B', 'A','B', 'A','B',], 
    'Store': ['S1','S1', 'S2' ,'S2' , 'S1' , 'S2' , 'S2' , 'S1'], 
    'Sales': [100 , 200 , 150 ,250 , 120 , 180 , 200 , 300] ,
    'Quantity': [10 , 15 ,12 , 18 , 8 , 20 , 15 , 25] , 
    'Date': pd.date_range('2023-01-01' , periods=8)
}
df = pd.DataFrame(data) 
df
```

```python
# Group by category and calculate the sum: 

cat = df.groupby('Category')['Sales'].sum()

cat 
```

```python
df
```

```python
# Group by stores and calculate the sum of the sales:

salesSumOfStore = df.groupby('Store')['Sales'].sum() 

salesSumOfStore
```

```python
# Group by multiple columns: 
# Group by Category and stores:

groupByCandS = df.groupby(['Category','Store'])['Sales'].sum()
groupByCandS
```
