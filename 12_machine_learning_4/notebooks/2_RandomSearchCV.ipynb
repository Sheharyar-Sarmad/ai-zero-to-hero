```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df['species'].value_counts()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop('species', axis=1)
y = df['species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5, random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model_knn = KNeighborsClassifier(n_neighbors=1)
```

```python
model_knn.fit(X_train, y_train)
```

```python
round(model_knn.score(X_train, y_train)*100,2)
```

```python
from sklearn.svm import SVC
```

```python
model_SVM = SVC(gamma='auto')
```

```python
model_SVM.fit(X_train, y_train)
```

```python
round(model_knn.score(X_train, y_train)*100,2)
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
classifier = RandomizedSearchCV(model_SVM, {
    'C': [1,10,20],
    'kernel': ['rbf','linear']
}, cv=5, return_train_score=False, n_iter=4)
```

```python
classifier.fit(X,y)
```

```python
classifier.cv_results_
```

```python
results = pd.DataFrame(classifier.cv_results_)
```

```python
results[['param_C','param_kernel','mean_test_score']]
```

```python
best_score = classifier.best_score_
best_model = classifier.best_estimator_
```

```python
y_pred_svm = best_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

test_accuracy = accuracy_score(y_test, y_pred_svm)

print("Best CV Score:", best_score)
print("Unseen Test Accuracy:", test_accuracy)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(confusion_matrix(y_test, y_pred_svm))
print(classification_report(y_test, y_pred_svm))
```
