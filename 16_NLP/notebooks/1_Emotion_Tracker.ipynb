# Label Encoding - Converting Text To Numbers

# What this code does:

> Finds all unique emotions in the dataset using .unique()

> Assigns a number (0, 1, 2, ...) to each unique emotion using a loop

> Replaces text labels with numbers using .map() so the model can understand them

# Why we do this:

> Machine Learning models work with numbers, not text - they can't process strings like "happy" or "sad"

> Keeps the original meaning - same emotions get the same number, different emotions get different number

```python
emotion_numbers: dict = {}
i: int = 0

for emotion in unique_emotions:
    emotion_numbers[emotion] = i
    i += 1

df['emotion'] = df['emotion'].map(emotion_numbers)
```

# Text Normalization - Lowercasing, Punctuation, Digit & ASCII Removal

# What this does:

> Converts all text to lowercase so the model treats similar words the same way. Removes punctuation marks so the model focuses on words instead of symbols. Removes all digits so the model focuses on text rather than numbers. Removes non-ASCII characters so the model only keeps standard English characters.

# Why we do this:

> Reduces vocabulary size and prevents the model from treating the same words differently. Removes unnecessary noise like punctuation, numbers, emojis, and special characters that don't add meaning for most NLP tasks. Standardizes text for consistent processing. Improves model performance by helping it focus on meaning rather than case, symbols, digits, or special characters.

```python
df['text'] = df['text'].apply(lambda x: x.lower())
```

```python
df['text'].head()
```

```python
import string

def remove_punctuations(text: str) -> str:
    if not text:                    # If empty
        return ""                   # Return empty string
    if not isinstance(text, str):   # If not string
        text = str(text)            # Convert it
    return text.translate(str.maketrans('', '', string.punctuation))
```

```python
# This is an example to understand the above function so get it

# Create a table that says: Replace 'a' with 'x', 'e' with '3'
table = str.maketrans('ae', 'x3')

text = "apple"
result = text.translate(table)
print(result)  # "xppl3"

# What happened?
# 'a' → 'x'
# 'p' → 'p' (no change)
# 'p' → 'p' (no change)
# 'l' → 'l' (no change)
# 'e' → '3'
```

```python
df['text'].apply(remove_punctuations) # Applying the function at text finally
```

```python
def remove_digit(text: str) -> str:
  if not text:
    return ""
  if not isinstance(text, str):
    text = str(text)
  return ''.join([i for i in text if not i.isdigit()])
```

```python
df['text'] = df['text'].apply(remove_digit)
```

```python
df['text'].head()
```

```python
def remove_emojis(text: str) -> str:
  if not text:
    return ""
  if not isinstance(text,str):
    text = str(text)
  return ''.join(filter(lambda x: x.isascii(), text))
```

```python
df['text'] = df['text'].apply(remove_emojis)
```

```python
def remove_stopwords(text: str) -> str:
  if not text:
    return ""
  if not isinstance(text, str):
    text = str(text)
  words = word_tokenize(text)
  return ''.join([word for word in words if word.lower() not in stop_words])
```

# Bag Of Words

> Bag of Words (BoW) is a technique used to convert text data into numerical vectors based on word counts — completely ignoring grammar and word order.



```python
df.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
vectorizer = CountVectorizer(
    max_features=1000,        # ← CRUCIAL! Keep only top 1000 words
    min_df=5,                 # ← Ignore rare words
    max_df=0.7,               # ← Ignore very common words
    stop_words='english',     # ← Remove stopwords
    lowercase=True
)
X = vectorizer.fit_transform(df['text'])
```

```python
# Convert to array (optional, for viewing)
X_array = X.toarray()

# Get feature names (vocabulary)
feature_names = vectorizer.get_feature_names_out()

# Create a DataFrame with the BoW features
bow_df = pd.DataFrame(X_array, columns=feature_names)

print("Vocabulary size:", len(feature_names))
print("BoW Matrix shape:", X.shape)
bow_df.head()
```

# Creating Model

> I will create on model not so much becasue its not a project i am teaching you NLP

```python
# BoW matrix
X = vectorizer.fit_transform(df['text'])
y = df['emotion']

# Now you can train models
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = MultinomialNB()
model.fit(X_train, y_train)

print("Accuracy:", model.score(X_test, y_test))
```

```python
# Reverse mapping
emotion_labels = {
    0: 'sadness',
    1: 'anger',
    2: 'love',
    3: 'surprise',
    4: 'fear',
    5: 'joy'
}

new_text = ["I feel so happy and excited today!"]
new_text_bow = vectorizer.transform(new_text)
prediction = model.predict(new_text_bow)[0]

print(f"Predicted emotion: {emotion_labels[prediction]}")
# Output: Predicted emotion: joy
```
