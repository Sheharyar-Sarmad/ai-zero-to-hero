# Import CountVectorizer Class

```python
# This is the basic import for CountVectorizer i explain it before as well, for more knowledge go to this
# https://scikit-learn.org/stable/modules/generated/sklearn.feature_extraction.text.CountVectorizer.html 
from sklearn.feature_extraction.text import CountVectorizer 
```

# Creating Dummy Corpus

```python
from typing import List

corpus: List[int]  = [
    "The quick brown fox jumps over the lazy dog.",
    "Machine learning is a subset of artificial intelligence.",
    "Natural language processing helps computers understand human language.",
    "Python is a popular programming language for data science.",
    "The weather today is sunny and warm with a light breeze.",
    "I enjoy reading books and drinking coffee in the morning.",
    "Artificial intelligence is transforming the healthcare industry.",
    "Deep learning models require large amounts of training data.",
    "The stock market showed significant growth this quarter.",
    "She played the piano beautifully at the concert last night.",
    "Climate change is a global challenge that needs urgent attention.",
    "The new smartphone has an amazing camera and long battery life.",
    "Students are preparing for their final exams this week.",
    "The restaurant serves delicious Italian pasta and pizza.",
    "Space exploration has always fascinated humanity for centuries."
]
```

# n Grams

```python
vectorizer: CountVectorizer = CountVectorizer(
    ngram_range=(2,2), # This means select two words each time default is 1,1 you can do whatever you want
    stop_words='english' # This removes a predefined list of common English words that
                         # usually don't carry much meaning (like "the", "is", "are", "and", etc.).
                         # These are called stop words.
)
```

```python
from scipy.sparse import spmatrix # This is optional but in production ready code you need typehints thats why i import
# sparse matrix, sparse means scattered thin, or not dense, in DataScience its generally means 0

X: spmatrix = vectorizer.fit_transform(corpus)
```

```python
print(f"Vocabulary\n: {vectorizer.get_feature_names_out()}")
print(f"\nBoW\n:{X.toarray()}")
```

# Making DataFrame

```python
import pandas as pd 

df: pd.DataFrame = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)
```
