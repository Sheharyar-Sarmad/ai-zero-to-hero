```python
# Broadcasting:
```

```python
# Broadcasting in NumPy is a powerful feature that lets you perform operations
# on arrays of different shapes without manually reshaping or looping.

#Example:

# Broadcasting with vector array: 

l = [10,20,30,40]
arr = np.array(l)
arr = arr + 10

arr

# Broadcasting allow you to perform arithmetic operation on a list by one specific given number
```

```python
# Broadcasting with matrix array: 

arrBroadMatrix = np.arange(1,26).reshape(5,5)
arrBroadMatrix = arrBroadMatrix + 10

arrBroadMatrix
```

```python
# Deep and Shallow Copy:
```

```python
# In NumPy, deep copy and shallow copy explain whether a new array is truly independent or still linked to the original.
# This is a very important concept because it controls whether changes affect original data or not.

a = np.arange(1,21)
a
```

```python
# This is shallow copy example:

slice = a[:5]
slice = slice * 10
slice 
```

```python
a[4:6] * 10
```

```python
a
```

```python
# This is deep copy example:
b = a
b[0] = 99
a
```

```python
# if you want that the changes will not occur on a then we can say:a

b = a.copy()
b[0] = 100
a[0] = 1
a 
```

```python
# Tip: You cannot stack two matrix arrays only vectors can get stacked:A

# Types of stacks:
# Vertical stack
# Horizontal stack
# Simple stack

# Example of vertical stack:

a = np.array([1,2,3,4])
b = np.array([5,6,7,8])

resultvstack = np.vstack((a,b)) # vtack() fuction only takes one positional 
resultvstack

```

```python
# Example of horizontal stack:

resulthstack = np.hstack((a,b))
resulthstack
```

```python
# Example of column stack:A

resultcstack = np.column_stack((a,b))
resultcstack
```
