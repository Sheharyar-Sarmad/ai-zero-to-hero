```python
import numpy as np
```

```python
# Numpy Indexing and Slicing of vectors:
```

```python
arr = np.arange(11,21)
arr
```

```python
# This is array "Indexing".
arr[6] # Simple formula of indexing and slicing is arr[start , stop , skip]
```

```python
# This is array "Slicing" 

# Array slicing means removing specific index value from an array.abs

arr[:] # This colun means start from zero index and go till end
```

```python
arr[:4] # This colun means start from zero and go till 4th
```

```python
arr[4:] # This colun means start from 4th index and go till end
```

```python
arr[0:5] # Array in numpy start from 0 index go until number of n and 
# when we say 0:5 start from zero and go till 5 but it will select select 0 from 4th element it will not go
# to 5th index it works like range function. 

# Simple rule:
# Select startingindex and endingIndex plus oneabs
# StartingIndex:EndingIndex+1  to get the exact ending index. 
```

```python
arr[3::2] # Now the third value means how much index you want to skip. 
```

```python
# Numpy Indexing and Slicing of matrix
```

```python
arrm = np.arange(1,31)
arrm = arrm.reshape(6,5)
arrm 
```

```python
arrm[0] # The first value in matrix indexing will target rows
```

```python
arrm[5] 
```

```python
arrm[0,4] # Now this will get the 0 row and 5 column element.
```

```python
arrm[5,4]
```

```python
arrm
```

```python
arrm[0:2,1:3]
```

```python
arrm[3:, 3: ]
```

```python
arrm[4: , 1:3]
```

```python
arrm
```

```python
arrm[2:4 , 1:3]
```

```python
arrm[:,2] # If you want to slice a full colomn then you should say like that 
```
