```python
# Data Frame: 

# A DataFrame is a 2-dimensional labeled data structure in Pandas that stores data in rows and columns (like a table in Excel).
# When we combine two or more series it becomes data frame. 
```

```python
# What we are gonna learn in it: 

# Creating data frames 
# Selection and indexing of columns 
# Creating a new column 
# Removing column 
# Removing row 
# Selecting row 
# Selecting subsets of rows and columns 
# Conditional selection  
```

```python
import numpy as np 
import pandas as pd 
```

```python
# Creating data frames: 

data = {
    'Name': ['John' , 'Anna' , 'Peter' , 'Linda'] ,
    'Age': [28 , 34 , 29 , 42] ,
    'City': ['New York' , 'Paris' , 'Berlin' , 'London'] ,
    'Salary': [65000 , 70000 , 62000 , 85000]    
}

df = pd.DataFrame(data)
```

```python
df 
```

```python
data_list = [
    ['John' , 28 , 'New York' , 65000],
    ['Anna' , 34 , 'Paris' , 70000],
    ['Peter' , 29 , 'Berlin' , 62000],
    ['Linda' , 42 , 'London' , 85000]
]

columns = ["Name", "Age", "City", "Salary"]
df2 = pd.DataFrame(data_list, columns=columns)
```

```python
df2
```

```python
# Selection and Indexing of Columns: 
```

```python
df2['Name'] # In indexing of columns in data frame didnt works the same like array index in data frame we have to name the column to get its data 
df2['City']
```

```python
df2[["Name" , "City"]] # If you want to get a two or multiple columns then create list of a list and then provide the columns names 
```

```python
# Creating a column inside a data frame:  

df2["Designation"] = ["Doctor" , "Software engineer" , "Doctor" , "Software engineer"] # This will create a new column at the end of the datafram 
# index and and df2["Designation"] means create a new column inside df2 data frame and then ["Doctor",...] means the values inside it be carefull 
# provide the exact number of values which all the columns have if you disturb the count it will throw an error.  
df2 
```

```python
# Removing a column inside a data frame 

df2.drop(['City' , 'Age'] , axis=1 , inplace=False)
```

```python
df2 
```

```python
# Removing rows:  
df2.drop(0,axis=0) # When you want to remove the row then do nothing just say its index value. 
```

```python
df2
```

```python
# Selection of the rows: 

# df2[0] # This will throw the keyerror. 

# Right way is: 

df2.loc[0] # loc means location and it will select 0 location index of the row 
```

```python
# If you want to select multiple rows then create list of list and provide the index: 

df2.loc[[0,3]]
```

```python
# Selecting subsets of rows and columns: 

df2.loc[[0 , 1],["City" , "Salary"]] # You need to provide the the index of dataframe inside a list and then in other list provide the column 
                                     # key. 

df2
```

```python
df2.loc[[2,3],["Name" , "Age"]]
```

```python
# Conditional selection: 

# Conditional selection means Selecting rows or data from a DataFrame based on a condition.

# for example i only want to see the people who's age is greater than 30: 
```

```python
df2[df2["Age"] > 30]
```

```python
# The age must be greater than 30 and the city must be paris only
```

```python
df2[(df2["Age"] > 30) & (df2["City"] == "Paris")]
```

```python
df2[df2["Salary"] < 80000]
```

```python
conditionalChecking = df2[ (df2["Salary"] < 70000) & (df2["Age"] < 30) ] # If you want to do and or and not operation then you 
# you just put each condition in paranthesis( () ). 
conditionalChecking
```
