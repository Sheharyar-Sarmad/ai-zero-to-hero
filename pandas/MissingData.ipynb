```python
# Missing Data: 

# Missing data means Values that are empty, unavailable, or not recorded in a DataFrame. 

# What we are going to study: 

# Finding missing data 
# Removing missing data 
# Filling missing data 
```

```python
import numpy as np 
import pandas as pd 
```

```python
data = {
    'A': [1 , 2 , np.nan , 4 , 5] , 
    'B': [np.nan , 2 , 3 , 4 , 5] ,
    'C': [1 , 2 , 3 , np.nan , np.nan] , 
    'D': [1 , np.nan , np.nan , np.nan , 5] ,
}

df = pd.DataFrame(data)
df
```

```python
# If we want to check that its number or we have to write function np.isna()

df.isna()
```

```python
# This will give you the count of null value inside each column 

df.isna().sum()
```

```python
# This will the rows and will answer you true if there's any null value in a column 

df.isna().any() # you can provide the axis=0 or axis=1 according to row or column need, 0 for rows and 1 for co. 
```

```python
# Removing Missing data from data frame: 

# Removing missing data means: Deleting rows or columns that contain missing values (NaN or None) from a DataFrame.
```

```python
# Apply this function if you want to remove all the nan from your data framework: 
df = pd.DataFrame(data)
df.dropna()
```

```python
df.dropna(thresh=3) # thresh means minimum number of non-missing (non-NaN) values required to keep a row or column
```

```python
# Filling missing data of a data frame: 

# Filling missing data means Replacing missing values (NaN or None) in a DataFrame with some meaningful value instead of deleting them.
```

```python
# for example: 

# If you want to fill all the na values of a data frame then do this: 

df.fillna(0)
```

```python
# for example: 

# If you want to fill the value of na column wise then do this: 

values = {'A':0,'B':100,'C':300,'D':400}
df.fillna(value=values)
```

```python
values2 = {'A':10,'B':20,'C':30,'D':40}
df.fillna(value=values2)
```
