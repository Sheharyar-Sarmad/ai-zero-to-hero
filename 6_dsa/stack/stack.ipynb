```python
class Stack:
    def __init__(self):
        self.container = deque()

    def push(self , val):
        self.container.append(val)
        
    def pop(self):
        return self.container.pop() 

    def peek(self): 
        return self.container[-1]
        
    def is_empty(self):
        return len(self.container)==0

    def size(self): 
        return len(self.container)

    def clear(self):
        return self.container.clear()
```

```python
s = Stack()

s.push(10)
s.push(20)
s.push(30)

print(s.pop())       # 30
print(s.size())      # 2
print(s.is_empty())  # False
print(s.clear())
print(s.is_empty())
```
