```python
class HashTable:
    def __init__(self):
        self.MAX = 10
        self.arr = [[] for i in range(self.MAX)]

    def get_hash(self, key):
        h = 0
        for char in key:
            h += ord(char)
        return h % self.MAX

    def __getitem__(self , key): 
       h = self.get_hash(key)

       for elem in self.arr[h]: 
           if elem[0] == key:
               return elem[1] 

    def __delitem__(self , key):
        h = self.get_hash(key)
        
        for idx,elem in enumerate(self.arr[h]): 
            if elem[0] == key:
                del self.arr[h][idx]
                break

    def __setitem__(self, key, val):
        h = self.get_hash(key)
        found = False

        # Search for the key in the bucket
        for idx, element in enumerate(self.arr[h]):
            if element[0] == key:
                self.arr[h][idx] = (key, val)
                found = True
                break

        # If key was not found, append a new tuple
        if not found:
            self.arr[h].append((key, val))

       
```

```python
t = HashTable()

print(t.get_hash("march 6"))
print(t.get_hash("march 17"))
```
