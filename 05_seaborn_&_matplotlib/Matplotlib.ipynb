```python
# Subplot in Matplotlib:
```

```python
# subplots are used when you want to create multiple plots
plt.subplot(2,2,1) # now you have to provide the rows and columns:
z = np.arange(1,11)
y = z + 100
plt.plot(z,y)
plt.subplot(2,2,2)
plt.plot(z,z**2)
plt.subplot(2,2,3)
plt.plot((z**2)-10 , (z**2)-10)
plt.subplot(2,2,4)
plt.plot(l,b)
```

```python
# data: 

g = np.arange(1,16)
h = g ** 3

plt.subplot(6,6,1)
plt.plot(g,h)

plt.subplot(6,6,2)
plt.plot(h-10,h+10)

plt.subplot(6,6,3)
plt.plot(g/2 , g/2)

plt.subplot(6,6,4)
plt.plot((h*2)/2 , (g*2)/2)

plt.subplot(6,6,5)
plt.plot((h*(9/5)+32) , (g*(9/5)+32))

plt.subplot(6,6,6)
plt.plot((g/h)*100 , (h/g)*100)
```

```python
# object oriented way: 

fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,0.6,0.6]) # 2. add_axes([left, bottom, width, height]) thats its concept

x = np.arange(1,21)
y = x ** 2

axis1.plot(x,y)
axis1.set_title("Square Function")
axis1.set_xlabel("X graph")
axis1.set_ylabel("Y graph")

axis2 = fig.add_axes([0.2,0.2,.3,.3])
axis2.plot(x,y)

# If we create two or more axis inside a figure they overlapp over each other 
```

```python
# practice of object oriented way(figure):

figure = plt.figure()

ax1 = figure.add_axes([.1,.1,.6,.6])

x = np.arange(1,11)
y = (x + 10) * 2

ax1.plot(x,y)
ax1.set_title("Practice axes 1")
ax1.set_xlabel("X graph")
ax1.set_ylabel("Y graph")

ax2 = figure.add_axes([.2,.2,.4,.4])
ax2.plot(y,x)
ax2.set_title("Practice axes 2")
ax2.set_xlabel("X graph")
ax2.set_ylabel("Y graph")
```
