```python
# basic imports: 

import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
# Categorical data plot: 

# Now lets use seaborn to plot categorical data! There are few types of main plot for this:

# boxplot
# violinplot
# stripplot
# swarmplot
# barplot
# countplot
```

```python
df = sns.load_dataset('tips')
df
```

```python
sns.countplot(x=df['sex'] , hue=df['smoker'])
```

```python
md = sns.load_dataset('geyser')

sns.countplot(x=md['kind'],hue=md['duration'])
```

```python
# bar plot:
```

```python
import numpy as np 
sns.barplot(x=df['sex'] , y=df['tip'] , estimator=np.sum) 
```

```python
sns.boxplot(x='tip' , y='day' , data=df , palette='rainbow')
```

```python
sns.violinplot(x='tip' , y='day' , data=df , palette='rainbow')
```

```python
sns.stripplot(x='day' , y='total_bill' , data=df)
```

```python
sns.swarmplot(x='day' , y='total_bill' , data=df)
```

```python
sns.violinplot(x='day' , y='tip' , data=df)
sns.swarmplot(x='day' , y='tip' , data=df)
```
