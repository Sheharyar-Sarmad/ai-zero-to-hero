```python
# JointPlot 

# A joint plot is a visualization that combines three plots in one figure to show the relationship between 
# two variables along with their individual distributions.

# It consists of:

# Component	Position	Purpose
# Central plot	Center (main area)	Shows the bivariate relationship (scatter, hexbin, kde, etc.)
# Top marginal plot	Above the central plot	Shows the univariate distribution of the x-variable
# Right marginal plot	To the right of central plot	Shows the univariate distribution of the y-variable
```

```python
sns.jointplot(x='total_bill' , y='tip' , data=df , kind='scatter')
# example total_bill first value is 10 and tip value is 12, so it will create pair of the values then add a dot
# on that point and so on till end and this will create a scatter and its used in ml algorithms and through this algorithm 
# we find the best-fit-line and this is called as linear-regression concept
```

```python
sns.jointplot(x='total_bill' , y='tip' , data=df , kind='reg')
```

```python
d = sns.load_dataset('flights')
sns.jointplot(x='year' , y='passengers' , data=d , kind='hex')
```

```python
# Rug plot: 

# A rug plot is a minimalist, one-dimensional visualization that shows the distribution of a single quantitative variable by
# drawing small vertical lines (called "rug ticks" or "rug marks") along an axis for each data point.
# It's essentially a 1D scatter plot compressed onto a line.
```

```python
sns.rugplot(df['total_bill'])
```
