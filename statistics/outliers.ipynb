```python
import numpy as np
arr = np.array([2,3,4,6,7,8,9,12,13,16,17,23,25,27,34,37,201])
arr
```

```python
# in numpy we have a function named as np.percentile for creating quartiles

Q1 = np.percentile(arr , 25)
Q3 = np.percentile(arr , 75)
```

```python
# Lets find IQR(Interquartile Range) now : 

IQR = Q3 - Q1
```

```python
UF = Q3 + (1.5 * IQR)
LF = Q1 - (1.5 * IQR)
```

```python
# This is the common and classic python way to set boundries of lower and upper fence : 
# l = []
# for i in arr :
#     if i >= LF and i <= UF :
#         l.append(i)
# arr2 = np.array(l)

# This is the efficient numpy way set upper and lower fence boundries
arr2 = arr[(arr >= LF) & (arr <= UF)]
arr2
```

```python

import seaborn as sns
sns.boxplot(x=arr2)
```
