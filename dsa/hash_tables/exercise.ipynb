```python
# nyc_weather.csv contains new york city weather for first few days in the month of January. 
# Write a program that can answer following,

# What was the temperature on Jan 9?
# What was the temperature on Jan 4?
```

```python
# 1
nyc_weather = {}
with open("nyc_weather.csv" , "r") as f: 
    next(f)
    for line in f:         
        token = line.split(",")
        date = token[0]
        temp = int(token[1].strip())
        nyc_weather[date] = temp
```

```python
# poem.txt Contains famous poem "Road not taken" by poet Robert Frost. You have to read this file in python
# and print every word and its count as show below. Think about the best data structure that you can use 
# to solve this problem and figure out why you selected that specific data structure.

 # 'diverged': 2,
 # 'in': 3,
 # 'I': 8
```

```python
poem = {}

with open("poem.txt" , 'r') as f:
    for line in f: 
        each_word = line.split()
        for word in each_word: 
            if word in poem: 
                poem[word] += 1
        else:
            poem[word] = 1

print(poem)
```
