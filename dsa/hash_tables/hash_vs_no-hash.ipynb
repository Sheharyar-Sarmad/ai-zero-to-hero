```python
stock_prices = []

with open("stock_prices.csv" , "r") as f:
    for line in f:
        tokens = line.split(",")
        day = tokens[0]
        price = float(tokens[1])
        stock_prices.append([day , price])
stock_prices
```

```python
for elements in stock_prices: # O(n) this is not an efficient way thats why hash table comes
    if elements[0] == 'march 9':
        print(elements[1])
```
