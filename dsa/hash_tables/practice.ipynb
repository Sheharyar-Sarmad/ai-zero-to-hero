```python
class HashTable:
    def __init__(self): 
        self.max_count_of_elem = 100
        self.arr_data = [None for i in range(self.max_count_of_elem)]

    def get_hash_value(self , key): 
        sum_of_hash = 0 

        for characters in key: 
            sum_of_hash += ord(characters)
        return sum_of_hash % self.max_count_of_elem

    def __setitem__(self , key , value): 
        hashed = self.get_hash_value(key)
        self.arr_data[hashed] = value

    def __getitem__(self , key):
        hashed = self.get_hash_value(key)
        return self.arr_data[hashed]

    def __delitem__(self , key): 
        hashed = self.get_hash_value(key)
        self.arr_data[hashed] = None

    
```
