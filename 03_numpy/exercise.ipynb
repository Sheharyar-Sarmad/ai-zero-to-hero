```python
import numpy as np 
```

```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],

    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],

    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
# axis = 0 always targets columns a
b = np.sum(s,axis=1)
b == 45

for i in b :
    if i != 45 :
        print("Sudoku is not valid!!")
        break
else :
    print("For rows its valid!!")
```

```python
c = np.sum(s,axis=0)
c == 45

for i in b :
    if i != 45 :
        print("Sudoku is not valid!!")
        break
else :
    print("For column its valid!!")
```

```python
# Column: [age , math marks , science marks] 
data = np.array([
    [18,85,78], #Student 1
    [19,92,88], #Student 2
    [17,76,95], #Student 3
    [18,65,70], #Student 4
    [20,90,85], #Student 5
])
```

```python
# Get the shape of the matrix: 

data.shape
```

```python
# Find the average age of students:

np.mean(data[:,0])
```

```python
# Extract maths marks of all students:

data[:,1]
```

```python
# Find the highest science marks: 

allScienceMarks = data[: , 2]
allScienceMarks.max()
```

```python
# Get details of the students who score 90 plus in math exams:

bool_marks = data[:,1] >= 90 
bool_marks
```

```python
# Increase Math marks of all students by 5:

mathdata = data[:,1]
result = mathdata + 5 
result
```

```python
# Find how many students are younger than 19:

len(data[data[:,0] < 19])
```

```python
# Calculate the average marks in each subject (column-wise): 

np.mean(data[:,1:] , axis=0)
```

```python
# Get the max of the students who score atleast 80 marks in both subject: 

data[ (data[:,1] >= 80) & (data[:,2] >= 80) ] 
```
