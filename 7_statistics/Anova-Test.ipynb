```python
import seaborn as sns
import pandas as pd 
from scipy.stats import f_oneway
```

```python
df = sns.load_dataset('titanic')
df = df[['age','pclass']].dropna()
```

```python
df['pclass'].unique()
```

```python
class_1 = df[df['pclass'] == 1]['age']
class_2 = df[df['pclass'] == 2]['age']
class_3 = df[df['pclass'] == 3]['age']
```

```python
f_stats , p_vlaue = f_oneway(class_1,class_2,class_3)
```

```python
f_stats
```

```python
p_vlaue
```

```python
alpha = 0.05

if p_vlaue < alpha:
    print("reject the null hypothesis becuase their is a significant difference")
else:
    print("there is no significant difference")
```
